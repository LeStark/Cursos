```python
import pandas as pd
import numpy as np

# -------------------------------
# 1️ Generador del dataset de DEMANDA
# -------------------------------

np.random.seed(42)

dias = pd.date_range(start="2024-01-01", end="2024-06-30", freq="D")  # 6 meses de datos
productos = [1001, 1002, 1003]

data_demanda = []

for p in productos:
    # Demanda base distinta por producto
    base = np.random.randint(15, 35)
    estacionalidad = 5 * np.sin(2 * np.pi * np.arange(len(dias)) / 7)
    ruido = np.random.normal(0, 4, len(dias))
    demanda = np.maximum(np.round(base + estacionalidad + ruido), 0).astype(int)
    
    for d, dem in zip(dias, demanda):
        data_demanda.append({
            "Fecha": d,
            "Producto": p,
            "Demanda": dem
        })

Data_demanda = pd.DataFrame(data_demanda)
print("📦 Data_demanda (primeras filas):")
print(Data_demanda.head())

# -------------------------------
# 2️ Generador del dataset de PEDIDOS (para cálculo de lead time)
# -------------------------------

np.random.seed(123)

# Supongamos que se hacen pedidos cada cierto tiempo para cada producto
Data_pedidos = []

for p in productos:
    # Cantidad de pedidos diferentes según producto
    n_pedidos = np.random.randint(6, 10)
    fechas_pedido = pd.to_datetime(
        np.random.choice(dias[:-10], n_pedidos, replace=False)
    )
    fechas_pedido = sorted(fechas_pedido)
    
    for f in fechas_pedido:
        # Lead time aleatorio entre 2 y 7 días
        lead_time = np.random.randint(2, 8)
        fecha_recepcion = f + pd.Timedelta(days=lead_time)
        cantidad = np.random.randint(300, 600)
        Data_pedidos.append({
            "Producto": p,
            "Fecha Pedido": f,
            "Fecha Recepción": fecha_recepcion,
            "Cantidad": cantidad
        })

Data_pedidos = pd.DataFrame(Data_pedidos)
print("\n📦 Data_pedidos (primeras filas):")
print(Data_pedidos.head())


Data_pedidos.to_csv("Data_pedidos.csv", index=False)
Data_demanda.to_csv("Data_demanda.csv", index=False)
```
